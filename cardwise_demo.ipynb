# CardWise AI — AI Credit Card Benefits Advisor
## Built with RAG, Hybrid Search (BM25 + FAISS), LangGraph, and LangSmith Observability

CardWise AI answers policy Q&A, performs head-to-head comparisons, and provides spend-based financial recommendations for the **Chase Sapphire Reserve** and **American Express Platinum Card**.

```python
# Set up path & environment
import sys
import os
from pathlib import Path

sys.path.insert(0, str(Path.cwd()))

from dotenv import load_dotenv
load_dotenv()

# LangSmith Observability optional setup
# os.environ["LANGCHAIN_TRACING_V2"] = "true"
# os.environ["LANGCHAIN_PROJECT"] = "CardWise-AI-Advisor"
```

### 1. Initialize CardWise Hybrid Retriever & LangGraph Workflow

```python
from src.graph import CardWiseGraph

print("Initializing CardWise Graph with BM25 + FAISS Hybrid Search...")
app = CardWiseGraph()
```

### 2. Scenario 1: Policy Q&A Lookup

```python
query_1 = "What is the cell phone protection policy for the Amex Platinum card, and what is the deductible?"
res_1 = app.run(query_1)

print(f"Intent: {res_1['intent']}")
print(f"Card Filter: {res_1['card_filter']}")
print("\n--- Answer ---")
print(res_1['final_answer'])
```

### 3. Scenario 2: Head-to-Head Comparison

```python
query_2 = "Compare airport lounge access and guest policies for Chase Sapphire Reserve vs Amex Platinum."
res_2 = app.run(query_2)

print(f"Intent: {res_2['intent']}")
print("\n--- Answer ---")
print(res_2['final_answer'])
```

### 4. Scenario 3: Financial Recommendation Engine

```python
query_3 = "I spend $12,000 on flights, $6,000 on hotels, and $8,000 on dining annually. Which card is better for me?"
spend_3 = {
    "flights_direct": 12000.0,
    "hotels_portal": 6000.0,
    "dining": 8000.0
}

res_3 = app.run(query_3, spend_profile=spend_3)
print(res_3['final_answer'])
```
